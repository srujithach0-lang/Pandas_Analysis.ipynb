Task 1 — Explore and Identify Issues Load the dataset and use df.info(), df.describe(), and df.shape to report at least three data quality problems you observe (e.g., wrong dtypes, nulls, impossible values).

```python
import pandas as pd

df = pd.read_csv("/content/used_cars_messy.csv")
df.shape
df.head()
df.info()
df.describe()
df.isnull().sum()

```

Task 2 — Clean the Data Fix the issues identified: drop null target rows, impute missing input features, strip and lowercase the brand column, extract numeric values from the mileage column, and remove duplicates

```python
import numpy as np  # <--- Add this line here to fix the NameError

# 1. Drop rows where the target variable (selling_price) is missing
df = df.dropna(subset=['selling_price'])

# 2. Strip whitespaces and lowercase the 'brand' column
df['brand'] = df['brand'].astype(str).str.strip().str.lower()

# 3. Extract numeric values from the 'mileage' column
df['mileage'] = df['mileage'].astype(str).str.extract(r'([\d\.]+)').astype(float)

# 4. Impute missing input features
num_cols = df.select_dtypes(include=[np.number]).columns
for col in num_cols:
    if col != 'selling_price':
        df[col] = df[col].fillna(df[col].median())

cat_cols = df.select_dtypes(include=['object']).columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# 5. Remove duplicate rows
df = df.drop_duplicates()

print(f"Cleaned dataset shape: {df.shape}")
```

Task 3 — Compute Baseline MAE Build a baseline model that predicts the mean selling_price for every record. Calculate and print the MAE of this baseline on the full cleaned dataset.

We will predict the mean of the selling_price for every single car and calculate the Mean Absolute Error ($MAE$).$$MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}|$$


```python
from sklearn.metrics import mean_absolute_error

# 1. Define actual values
y_true = df['selling_price']

# 2. Create baseline predictions (the mean of selling_price for every row)
mean_price = y_true.mean()
y_pred_baseline = [mean_price] * len(df)

# 3. Calculate Mean Absolute Error
baseline_mae = mean_absolute_error(y_true, y_pred_baseline)

print(f"Baseline Mean Price: {mean_price:.2f}")
print(f"Baseline Mean Absolute Error (MAE): {baseline_mae:.2f}")
```
