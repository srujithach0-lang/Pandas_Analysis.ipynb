```python
import pandas as pd
import numpy as np
# 1. Initialize the synthetic dataset
df = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 2, 5],
    'age':         [25, np.nan, 35, np.nan, np.nan, 40],
    'city':        ['Mumbai', 'Delhi', np.nan, 'Mumbai', 'Delhi', np.nan],
    'gender':      ['Male', 'Female', 'Female', 'Male', 'Female', 'Male'],
    'annual_income': [40000, 80000, 60000, 120000, 80000, 95000]
})
```

# Task 1 — Clean the Data
Using the synthetic dataset below, write Python code to:

Impute missing values in age with the median and in city with the mode
Remove any duplicate rows



```python
#Calculate the median of 'age'
median_age = df['age'].median()

#calculaing the mode of 'city'

mode_city = df['city'].mode()

#impute the missing values
#age with median
df['age'] = df['age'].fillna(median_age)
df['city'] = df['city'].fillna(mode_city)

print(median_age)
print(mode_city)

```

```python
#Removing duplicate values
df_Cleaned = df.drop_duplicates()
df.drop_duplicates(inplace=True)
```

```python
# Check how many duplicate rows exist
print(df.duplicated().sum())

# View the duplicate rows
print(df[df.duplicated()])

# Remove duplicates, keep the first occurrence
df.drop_duplicates(inplace=True)
```

# Task 3 — Scale Numerical Features
Scale age and annual_income using both methods below on the cleaned dataframe, and print the results of each:
Min-Max Scaling (using MinMaxScaler)

In 2–3 sentences, explain when you would prefer one over the other.

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df[["age", "annual_income"]] = scaler.fit_transform(df[["age", "annual_income"]])
```

# Task 3 Explanation: MinMaxScaler vs. StandardScaler
➡️Prefer MinMaxScaler when your data does not follow a normal distribution, you have strictly bounded limits, or your algorithm specifically requires features to be on an exact scale between 0 and 1 (like neural networks or distance-based algorithms like KNN).

➡️Prefer StandardScaler when your data follows a normal (Gaussian) distribution or contains extreme outliers, because standardisation does not compress outliers into a tiny, squished range like Min-Max scaling does. Many machine learning estimators (like linear regression, logistic regression, and SVMs) assume that all features are centered around zero with a variance of one.
